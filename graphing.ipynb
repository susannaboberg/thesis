## Second Run (num_samples=45)

```python
import json, matplotlib.pyplot as plt
s = json.load(open("runs/run2/batch_1_power_samples.json"))["gpu"]
t0 = s[0]["timestamp_s"]
t = [x["timestamp_s"] - t0 for x in s]
w = [x["power_w"][0] for x in s]
plt.plot(t, w)
plt.axhline(170.8, ls="--", c="gray", label="mean 170.8 W")
plt.xlabel("Time (s)"); plt.ylabel("GPU power (W)")
plt.title("RTX 4090 · Qwen2.5-0.5B · 45 responses"); plt.legend(); plt.show()
```
